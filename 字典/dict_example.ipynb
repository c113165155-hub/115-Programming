# 字典

字典，在Python中用於查表。
- 字典的前後是用大括號{}包住
- 裡面每個元素用逗點(,)隔開
- 每個元素都是key.value的形式


```python
students={'Peter':175,'Mary':168,'Kevin':165}
# 查學生表格，使用關鍵字(key) Kevin來查
print(students['Kevin'])
```

```python
students={
    'Peter':[175,70,'高雄'],
    'Mary':[168,55,'嘉義'],
    'Kevin':[165,55,'高雄']
}
print(students['Mary'])
```

```python
print('Peter' in students)
print('peter' in students) # 大小寫敏感
print('JoJo' in students)
```

```python
#新增 John 到 students 中
students['John']=[180,99,'台北']
```

```python
# 由students 中移除 Kevin
del students['Kevin']
print(students)
```

```python
# len可以得到字典中元素的個數
print(len(students))
```
