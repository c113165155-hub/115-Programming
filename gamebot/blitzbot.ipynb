# 介紹

本程式使用pyautogui來完成一個Blitz的遊戲機器人

```python
import pyautogui as pyg
pyg.sleep(2)
basex=512
basey=383
gap=40
```

```python
for row in range(8):
    for col in range(2):
        x=basex+col*gap
        y=basey+row*gap
        color=pyg.pixel(x,y)
        print(f"({row},{col}) at ({x},{y}):{color}")
```
