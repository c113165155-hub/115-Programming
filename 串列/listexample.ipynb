# 簡介

串列是一個可以儲存多個元素 (整數、浮點數、或字串) 的容器。

串列中每個元素都是使用逗點隔開，左右有中括號[]。

```python
#score是一個串列，裡面每個元素都是整數。
score=[78,29,56,80,99]
print(score)
```

## 索引值

我們可以使用索引值(index)來存取串列中的元素。
注意:索引值由0開始。

```python
print(score[0]) #印出第1個元素
print(score[1]) #印出第2個元素
print(score[2]) #印出第3個元素
```

## 串列內容的個數

len(a)取得a這個串列中元素的個數
len是length(長度)的意思

```python
print(len(score))
print(score[4]) #取得最後一個元素
print(score[5-1]) #取得最後一個元素
print(score[len(score)-1])
print(score[-1])
```
